```python
from embeddings import embed_text
from vector_store import FaissStore
```

```python
store = FaissStore()

print("Total Documents:", store.count())
```

```python
query = "What is FoodLens AI?"

query_vector = embed_text(query)

results = store.search(
    query_vector,
    top_k=5
)

results
```

```python
for result in results:
    print("=" * 50)

    print("Title:",
          result["metadata"].get("title"))

    print("Type:",
          result["metadata"].get("type"))

    print("Score:",
          result["score"])
```

```python
query = "Who developed FoodLens AI?"
```

```python
query_vector = embed_text(query)

results = store.search(
    query_vector,
    top_k=5
)

for result in results:
    print(result["metadata"]["title"])
```
